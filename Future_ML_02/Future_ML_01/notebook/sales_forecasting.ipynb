```python
monthly_data = df.groupby(['Year', 'Month', 'Category', 'Region']).agg({
    'Quantity': 'sum',
    'Discount': 'mean',
    'Profit': 'sum',
    'Sales': 'sum'
}).reset_index()

monthly_data = monthly_data.sort_values(
    ['Year', 'Month', 'Category', 'Region']
)

monthly_data['Previous_Month_Sales'] = monthly_data.groupby(
    ['Category', 'Region']
)['Sales'].shift(1)

monthly_data = monthly_data.dropna()

monthly_data.head()
```

```python
X = monthly_data[[
    'Quantity',
    'Discount',
    'Profit',
    'Category',
    'Region',
    'Month',
    'Previous_Month_Sales'
]]
y = monthly_data["Sales"]
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder

categorical_features = ["Category", "Region"]

preprocessor = ColumnTransformer(
    transformers=[
        (
            "cat",
            OneHotEncoder(handle_unknown="ignore"),
            categorical_features
        )
    ],
    remainder="passthrough"
)
```

```python
split = int(len(monthly_data) * 0.8)

X_train = X.iloc[:split]
X_test = X.iloc[split:]

y_train = y.iloc[:split]
y_test = y.iloc[split:]
```

```python
from sklearn.pipeline import Pipeline
from sklearn.linear_model import LinearRegression

model = Pipeline([
    ("preprocessor", preprocessor),
    ("regressor", LinearRegression())
])

model.fit(X_train, y_train)
```

```python
y_pred = model.predict(X_test)

print(y_pred)
```

```python
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
import numpy as np

mae = mean_absolute_error(y_test, y_pred)

rmse = np.sqrt(mean_squared_error(y_test, y_pred))

r2 = r2_score(y_test, y_pred)

print("MAE:", mae)
print("RMSE:", rmse)
print("R² Score:", r2)
```

```python
results = pd.DataFrame({
    "Actual_Sales": y_test.values,
    "Predicted_Sales": y_pred
})

results.head(10)
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(10, 5))

plt.plot(results["Actual_Sales"], label="Actual Sales")
plt.plot(results["Predicted_Sales"], label="Predicted Sales")

plt.xlabel("Test Data")
plt.ylabel("Sales")
plt.title("Actual vs Predicted Sales")
plt.legend()
plt.show()
```

```python
from sklearn.ensemble import RandomForestRegressor
rf_model = Pipeline([
    ("preprocessor", preprocessor),
    ("regressor", RandomForestRegressor(
        n_estimators=200,
        random_state=42
    ))
])

rf_model.fit(X_train, y_train)
```

```python
rf_pred = rf_model.predict(X_test)

print(rf_pred)
```

```python
rf_mae = mean_absolute_error(y_test, rf_pred)

rf_rmse = np.sqrt(mean_squared_error(y_test, rf_pred))

rf_r2 = r2_score(y_test, rf_pred)

print("Random Forest MAE:", rf_mae)
print("Random Forest RMSE:", rf_rmse)
print("Random Forest R²:", rf_r2)
```

```python
comparison = pd.DataFrame({
    "Model": ["Linear Regression", "Random Forest"],
    "MAE": [mae, rf_mae],
    "RMSE": [rmse, rf_rmse],
    "R²": [r2, rf_r2]
})

comparison
```

By the above comparision the best model from the both is to be Random forest because R2 is higher and mae and rmse is lower for Random forest model so choosing it is more better 

```python
if rf_rmse < rmse:
    best_model = rf_model
    best_pred = rf_pred
    best_model_name = "Random Forest"
else:
    best_model = model
    best_pred = y_pred
    best_model_name = "Linear Regression"

print("Best Model:", best_model_name)
```

```python
final_results = pd.DataFrame({
    "Actual Sales": y_test.values,
    "Predicted Sales": best_pred
})

final_results.head(10)
```

```python
plt.figure(figsize=(10, 5))

plt.plot(final_results["Actual Sales"], label="Actual Sales")
plt.plot(final_results["Predicted Sales"], label="Predicted Sales")

plt.xlabel("Test Period")
plt.ylabel("Sales")
plt.title(f"{best_model_name} - Actual vs Predicted Sales")
plt.legend()
plt.show()
```
